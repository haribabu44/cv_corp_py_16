```
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __str__(self):
        return f'name:{self.name} age:{self.age}'
s1=Student("Hari",13)
print(s1.__str__())
print(s1)
print(str(s1))
```

```

class Car:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def __str__(self):
        return f"{self.brand} costs {self.price}"


c1 = Car("BMW", 5000000)

print(c1)
```

```
class Car:
    def __init__(self,brand,price):
        self.brand=brand
        self.price=price
    def __str__(self):
        return f'{self.brand} costs {self.price}'
    def __repr__(self):
        return f'Car({self.brand},{self.price})'
c1=Car("bmw",123255)
print(c1)
print(repr(c1))
print(f'{c1}')
```

```
class Add:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return f'{self.x+other.x},{self.y+other.y}'
a1=Add(1,2)
a2=Add(3,4)
print(a1+a2)
```

```
class Student:
    def __init__(self,marks):
        self.marks=marks
    def __add__(self,o2):
        return self.marks+o2.marks
    def __sub__(self,o2):
        return self.marks-o2.marks
s1=Student(90)
s2=Student(46)
print(s1+s2)
print(s1-s2)
```

```
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def __lt__(self, other):
        return self.marks<other.marks
s1=Student("hari",89)
s2=Student("mani",56)
print(s1<s2)
print(s2<s1)
```
