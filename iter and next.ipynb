```python
#using __Iter__ and __next__ find vowels in the string

class Vowels:
    def __init__(self,s):
        self.s=s
        self.i=0
    def __iter__(self):
        return self
    def __next__(self):
        while self.i<len(self.s):
            i=self.i
            self.i+=1
            if self.s[i] in "aeiou":
                return self.s[i]
        raise StopIteration






v1=Vowels("JustThinking")
it=iter(v1)
print(next(it))
print(next(it))
print(next(it))
```

```python
#custom iterator to find  even numbers in the list


class Even:
    def __init__(self,l):
        self.l=l
        self.index=0
        self.count=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index<len(self.l):
            i=self.index
            self.index+=1
            if self.l[i]%2==0:
                self.count=self.count+1
                return self.count
                # return self.l[i]
        else:
            raise StopIteration
l=Even([1,2,4,909,78,55,25,67,889,65,45,44])
for i in l:
    if i is not None:
        print(i)
```

```python
#by using custom iterators finding first n even numbers

class Customiterator:
    def __init__(self, n):
        self.n = n
        self.count = 0
        self.start = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.n:
            self.i = self.start
            self.start += 2
            self.count += 1
            return self.i
        raise StopIteration


v1 = Customiterator(10)
it = iter(v1)

for i in it:
    print(i)
```
