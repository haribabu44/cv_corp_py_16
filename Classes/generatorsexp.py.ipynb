```python
def fun(l):
    while True:
        sum=0
        for i in l:
            sum=sum+i
            yield sum

gen=fun([1,2,3,4,5,6])
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
def square(l):
    for i in l:
        k=i*i
        yield k



gen=square([2,3,4,5,6,7])
# print(next(gen))
# print(next(gen))
# print(next(gen))
for i in gen:
    print(i)
```

```python
def vowels(str):
    for i in str:
        if i in 'a,e,i,o,u':
            yield i




gen=vowels("haribabu")
# print(next(gen))
# print(next(gen))
# print(next(gen))
# print(next(gen))
for i in gen:
    print(i)
```

```python
def digit(str):
    for i in str:
        if i.isdigit():
            yield i





gen=digit("hari123")
# print(next(gen))
# print(next(gen))
# print(next(gen))
for i in gen:
    print(i)
```
