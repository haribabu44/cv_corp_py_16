```python
class Car:
    wheels=4
    def __init__(self,m):
        self.mileage=m
    def display_specs(self):
        print(f'mileage={self.mileage}')
        print(f'wheels={self.wheels}')
    @classmethod
    def change(cls,nw):
        cls.wheels=nw
        print(f'wheels={cls.wheels}')
c1=Car(15)
c1.display_specs()
c2=Car(20)
c2.change(8)
c1.display_specs()
c2.display_specs()
```

```python
class Employee:
    bonus_rate=0.1
    def __init__(self,n,base_sal):
        self.name=n
        self.base_salary=base_sal
    def final_salary(self):
        k=self.base_salary+(self.base_salary*self.bonus_rate)
        print(f'final_salary={k}')
    @classmethod
    def update_bonus(cls,new_rate):
        cls.bonus_rate=new_rate
    @staticmethod
    def is_valid(base_sal):
        return base_sal>1000
e1=Employee("hari",100000)
e1.final_salary()
e1.update_bonus(0.5)
e1.final_salary()
```
