```python
#comprehensive list

l=[i for i in range(10)]
print(*l)
```

```python
#comprehensive if

l=[i for i in range(10) if i%2==0]
print(*l)
```

```python
#comprehensive if - else

l=[i if i%2==0 else i+5 for i in range(10)]
print(*l)
```

```python

#comprehensive Zip



l1=[10,20,30,40]
l2=[50,60,70,80]

l=[i+j for i,j in zip(l1,l2)]
print(*l)


```

```python
# comprehensive nested if
l=["12345","hello@1765","12345678"]
d={ i:"strong password" if len(i)>8
    else "password" if len(i)==8
    else "weak password" for i in l}
print(d)
```

```python
n=[1,2,3,4,5,6,7]
l=[i*i for i in n]
print(*l)
```

```python
marks=[35,45,67,89,46,12]
l=["pass" if i>=40 else "fail" for i in marks]
print(*l)
```

```python
#4. Given a list containing duplicate numbers, create a set containing only the unique even numbers using set comprehension.
numbers = [10, 15, 10, 20, 25, 20, 30, 35, 30]
l={i for i in numbers if i%2==0}
print(*l)
```

```python
#5. Given a list of words, create a set containing the lengths of the words using set comprehension.
words = ["Python", "Java", "C", "Django", "Python"]
l={len(i) for i in words}
print(*l)
```

```python
#Create a new dictionary using dictionary comprehension where each student's name is mapped to "Pass" or "Fail".
students = {
    	    "Rahul": 75,
    	    "Anil": 32,
    	    "Priya": 56,
    	    "Sneha": 28
    	    }
l={i:"pass" if j>35 else "fail" for i,j in students.items()}
print(l)
```

```python
usernames = ["charan", "rahul", "priya", "sneha"]
passwords = ["abc123", "xyz456", "pqr789", "hello123"]
l={i:j for i,j in zip(usernames,passwords) }
print(l)
```

```python
#Create a new dictionary where the value is "Expensive" if the price is greater than 10000, otherwise "Affordable".
products = {
    	    "Laptop": 65000,
    	    "Mouse": 500,
    	    "Keyboard": 1500,
    	    "Monitor": 12000
    	    }
l={i:"Expensive" if j>10000 else "Affordable" for i,j in products.items()}
print(l)
```

```python
#Create a new dictionary containing only students who passed, using dictionary comprehension.
students = {"Rahul": 35,"Anil": 72,"Priya": 81,"Sneha": 29,"Kiran": 65}
l={i:j for i,j in students.items() if j>35}
print(l)
```
