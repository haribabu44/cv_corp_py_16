```python
#comprehensive list

l=[i for i in range(10)]
print(*l)
```

```python
#comprehensive if

l=[i for i in range(10) if i%2==0]
print(*l)
```

```python
#comprehensive if - else

l=[i if i%2==0 else i+5 for i in range(10)]
print(*l)
```

```python

#comprehensive Zip



l1=[10,20,30,40]
l2=[50,60,70,80]

l=[i+j for i,j in zip(l1,l2)]
print(*l)


```

```python
# comprehensive nested if
l=["12345","hello@1765","12345678"]
d={ i:"strong password" if len(i)>8
    else "password" if len(i)==8
    else "weak password" for i in l}
print(d)
```
